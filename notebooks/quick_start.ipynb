# CryptoGuardLite - Quick Start Tutorial

This notebook demonstrates the basic usage of CryptoGuardLite for encrypted traffic analysis.

```python
import sys
sys.path.insert(0, '../src')

import numpy as np
import torch
from models import BaselineCNN, HybridModel
from feature_extraction import TrafficFeatureExtractor
from data_processing import DataProcessor
from utils import config

print("CryptoGuardLite modules loaded successfully!")
```

## 1. Configuration

```python
# Check configuration
print("Attack Types:")
for attack_type in config.get('attack_types', []):
    print(f"  - {attack_type}")

print("\nModel Configuration:")
print(f"  CNN Filters: {config.get('model.hybrid.cnn_filters')}")
print(f"  LSTM Hidden Size: {config.get('model.hybrid.lstm_hidden_size')}")
print(f"  Transformer Heads: {config.get('model.hybrid.transformer_heads')}")
```

## 2. Feature Extraction

```python
# Initialize feature extractor
extractor = TrafficFeatureExtractor(
    max_packet_length=1500,
    sequence_length=100
)

print(f"Feature Extractor initialized")
print(f"  Max Packet Length: {extractor.max_packet_length}")
print(f"  Sequence Length: {extractor.sequence_length}")
```

## 3. Model Initialization

```python
# Initialize baseline CNN model
baseline_model = BaselineCNN(input_dim=100, num_classes=10)
print("Baseline CNN Model:")
print(f"  Input Dim: {baseline_model.input_dim}")
print(f"  Num Classes: {baseline_model.num_classes}")

# Initialize hybrid model
hybrid_model = HybridModel(
    input_dim=100,
    feature_dim=5,
    num_classes=10,
    cnn_filters=[64, 128, 256],
    lstm_hidden_size=128,
    transformer_heads=8,
    transformer_layers=4
)
print("\nHybrid Model:")
print(f"  Input Dim: {hybrid_model.input_dim}")
print(f"  Num Classes: {hybrid_model.num_classes}")
```

## 4. Test Model Forward Pass

```python
# Create dummy input
batch_size = 4
sequence_length = 100
feature_dim = 5

dummy_input = torch.randn(batch_size, sequence_length, feature_dim)

# Test baseline model
baseline_model.eval()
with torch.no_grad():
    baseline_output = baseline_model(dummy_input)

print(f"Baseline Model Output Shape: {baseline_output.shape}")

# Test hybrid model
hybrid_model.eval()
with torch.no_grad():
    hybrid_output = hybrid_model(dummy_input)

print(f"Hybrid Model Output Shape: {hybrid_output.shape}")
```

## 5. Data Processing

```python
# Initialize data processor
data_processor = DataProcessor()

# Generate dummy data
num_samples = 100
features = np.random.randn(num_samples, sequence_length, feature_dim)
labels = np.random.randint(0, 10, num_samples)

# Split data
X_train, X_test, y_train, y_test = data_processor.train_test_split(
    features, labels, test_size=0.2
)

print(f"Training set size: {len(X_train)}")
print(f"Test set size: {len(X_test)}")

# Create data loader
train_loader = data_processor.create_dataloader(
    X_train, y_train, batch_size=16, shuffle=True
)

print(f"Number of batches: {len(train_loader)}")
```
